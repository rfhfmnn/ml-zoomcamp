## Accessing elements

```python
df.loc[[1,2]] # extraer las filas 1 y 2 del dataframe df (a través de una lista)
```

```python
df.index=['a','b','c','d','e'] #cambiar el índice del dataframe df
df
```

```python
df.loc[['b','c']] # extraer las filas con índices 'b' y 'c' del dataframe df (a través de una lista)
```

```python
df.iloc[[1,2,4]] # extraer las filas 1, 2 y 4 del dataframe df (a través de una lista)
```

```python
df.reset_index(drop=True) #resetear el índice del dataframe df. drop=True indica que no queremos mantener la columna del índice anterior en el dataframe resultante.
```
