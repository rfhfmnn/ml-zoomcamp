```python
np.random.seed(2) #establece la semilla para generar números aleatorios
np.random.rand(5,2) # crea un array de 5 filas y 2 columnas con valores aleatorios entre 0 y 1 y con distribución uniforme
```

```python
np.random.seed(2) #establece la semilla para generar números aleatorios
np.random.randn(5,2) # crea un array de 5 filas y 2 columnas con valores aleatorios entre 0 y 1 y con distribución normal
```

```python
np.random.seed(2) #establece la semilla para generar números aleatorios
100*np.random.randn(5,2) # crea un array de 5 filas y 2 columnas con valores aleatorios entre 0 y 1 y con distribución normal
```

```python
np.random.seed(2) #establece la semilla para generar números aleatorios
np.random.randint(low=1, high=100, size=(5,2)) # crea un array de 5 filas y 2 columnas con valores aleatorios enteros entre 1 y 10
```

### Element-wise operations

```python
a=np.arange(5) #crea un array con valores del 0 al 4
a
```

```python
b=(10+(a*2))**2/100 #crea un array con valores del 0 al 4, los multiplica por 2, le suma 10, eleva al cuadrado y divide entre 100
b
```

```python
a+b
```

### Operaciones de comparación

```python
a>=1 # compara cada elemento del array a con el valor 1 y devuelve un array de booleanos
```

```python
a>b
```

```python
a[a>b] # devuelve un array con los elementos de a que son mayores que los elementos de b
```
