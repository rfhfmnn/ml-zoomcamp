#### multiplicación

```python
def vec_vec_mult(u,v):
    assert u.shape[0] == v.shape[0] # ver si los vectores tienen la misma dimension
    n=u.shape[0] # obtener la longitud de los vectores
    result = 0.0 # inicializar el resultado
    for i in range(n):
        result=result+u[i]*v[i] # multiplicar los elementos correspondientes y sumarlos
    return result
```

```python
 vec_vec_mult(u,v)
```

```python
u.dot(v)
```

```python
U=np.array([[1,2,3],[4,5,6]])
```

```python
U
```

```python
def mat_vex_mult(U,v):
    assert U.shape[1] == v.shape[0] # ver si la segunda dimension de U (columnas) es igual a la primera dimension de v (filas)
    num_rows=U.shape[0] # obtener el numero de filas de U
    result=np.zeros(num_rows) # inicializar el resultado
    for i in range(num_rows):
        result[i]=vec_vec_mult(U[i],v) # multiplicar la fila i de U con el vector v
    return result
```

```python
mat_vex_mult(U,v)
```

```python
U.dot(v)
```

```python
V=np.array([[1,2],[3,4],[5,6]])
```

```python
def mat_mat_mult(U,V):
    assert U.shape[1] == V.shape[0] # ver si la segunda dimension de U (columnas) es igual a la primera dimension de V (filas)
    num_rows=U.shape[0] # obtener el numero de filas de U
    num_cols=V.shape[1] # obtener el numero de columnas de V
    result=np.zeros((num_rows,num_cols)) # inicializar el resultado
    for i in range(num_cols):
        vi=V[:,i] # obtener la columna i de V
        Uvi=mat_vex_mult(U,vi) # multiplicar U con la columna i de V
        result[:,i]=Uvi # asignar el resultado a la columna i de result
    return result
```

```python
mat_mat_mult(U,V)
```

```python
U.dot(V)
```
