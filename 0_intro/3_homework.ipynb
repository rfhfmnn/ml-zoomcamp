```python
X=np.array(car.loc[car['origin']=='Asia',['vehicle_weight','model_year']].head(7))
X
```

```python
XT=X.T
XT
```

```python
XTX=XT.dot(X)
XTX
```

```python
Y=np.array([1100, 1300, 800, 900, 1000, 1100, 1200])
Y
```

```python
XTX_inv=np.linalg.inv(XTX)
XTX_XT=XTX_inv.dot(XT)
w=np.round(XTX_XT.dot(Y),2)
w
```
